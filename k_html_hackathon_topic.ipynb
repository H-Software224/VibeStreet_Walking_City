```python
df_1 = SEOUL_DONGDAEMUNGO_PARKINFORMATION_20240617.merge(SEOUL_DONGDAEMUNGO_PARKTREEINFORMATION_20221212, how='left', left_on='공원통합명', right_on='공원명')
df_1.head()
```

```python
df_2 = SEOUL_DONGDAEMUNGO_CROSSINFORMATION_20221212.merge(SEOUL_DONGDAEMUNGO_PARKTREEINFORMATION_20221212, how='left', left_on='지번주소', right_on='지번주소')
df_2.head()
```

```python
df_1 = df_1.drop(['공원면적','지번주소', '공원통합명'], axis=1)
df_2 = df_2.drop(['도로명', '도로시작점', '도로종료점', '수목위치'], axis=1)
df = pd.concat([df_1, df_2])
df.head()
df.to_csv(f'{FOLDER_LOCATION}/UPDATE_DATASET/UPDATED_DB.csv', encoding='cp949')
```
